```python
# Simple Federated Learning for House Price Prediction

import numpy as np
from sklearn.linear_model import LinearRegression

# -------- CLIENT DATA --------
X1 = np.array([[1000], [1200], [1500]])
y1 = np.array([50, 60, 75])

X2 = np.array([[1100], [1400], [1600]])
y2 = np.array([55, 70, 80])

# -------- CLIENT TRAINING --------
client1 = LinearRegression()
client1.fit(X1, y1)

client2 = LinearRegression()
client2.fit(X2, y2)

# -------- GET LOCAL WEIGHTS --------
w1 = client1.coef_[0]
w2 = client2.coef_[0]

# -------- FEDERATED AVERAGING --------
global_weight = (w1 + w2) / 2

print("Client 1 Weight:", w1)
print("Client 2 Weight:", w2)
print("Global Weight:", global_weight)
```
