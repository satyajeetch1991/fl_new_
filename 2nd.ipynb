```python
# Simple Federated Learning Data Preprocessing

import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
import numpy as np

# -------- HARD CODED DATASET --------
data = pd.DataFrame({
    'study_hours': [5, 7, 4, 6],
    'attendance': [80, 90, 70, 85],
    'internal_marks': [25, 30, 20, 28],
    'final_score': [60, 75, 55, 70]
})

print("Dataset:\n")
print(data)

# -------- FEATURES AND LABEL --------
X = data[['study_hours', 'attendance', 'internal_marks']]
y = data['final_score']

# -------- NORMALIZATION --------
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# -------- TRAIN TEST SPLIT --------
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# -------- CREATE FEDERATED CLIENTS --------
client1_X = X_train[:1]
client2_X = X_train[1:2]

print("\nClient 1 Data:")
print(client1_X)

print("\nClient 2 Data:")
print(client2_X)
```
