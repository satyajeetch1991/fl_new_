```python
# Simple Federated Learning in One File

import numpy as np

# -------- CLIENT --------
class Client:
    def __init__(self, data, label):
        self.data = data
        self.label = label
        self.weight = np.random.rand()

    def train(self):
        prediction = self.data * self.weight
        error = prediction - self.label

        # Weight update
        self.weight = self.weight - 0.01 * error.mean()

        return self.weight


# -------- SERVER --------
class Server:
    def federated_average(self, weights):
        return np.mean(weights)


# -------- CLIENT DATA --------
data1 = np.array([1, 2, 3, 4])
label1 = np.array([2, 4, 6, 8])

data2 = np.array([2, 3, 4, 5])
label2 = np.array([4, 6, 8, 10])

# -------- CLIENT TRAINING --------
client1 = Client(data1, label1)
client2 = Client(data2, label2)

w1 = client1.train()
w2 = client2.train()

print("Client 1 Weight:", w1)
print("Client 2 Weight:", w2)

# -------- SERVER AGGREGATION --------
server = Server()
global_weight = server.federated_average([w1, w2])

print("Global Weight:", global_weight)
```
