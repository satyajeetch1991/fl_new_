```python
# Simple Federated Learning using Logistic Regression

from sklearn.linear_model import LogisticRegression
import numpy as np

# -------- CLIENT 1 DATA --------
X1 = [[1], [2], [3], [4]]
y1 = [0, 0, 1, 1]

# -------- CLIENT 2 DATA --------
X2 = [[2], [3], [4], [5]]
y2 = [0, 1, 1, 1]

# -------- LOCAL TRAINING --------
client1 = LogisticRegression()
client1.fit(X1, y1)

client2 = LogisticRegression()
client2.fit(X2, y2)

# -------- GET WEIGHTS --------
w1 = client1.coef_[0][0]
w2 = client2.coef_[0][0]

print("Client 1 Weight:", w1)
print("Client 2 Weight:", w2)

# -------- FEDERATED AVERAGING --------
global_weight = (w1 + w2) / 2

print("Global Weight:", global_weight)
```
