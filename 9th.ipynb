```python
# Simple Federated Learning with Differential Privacy

import numpy as np
from sklearn.linear_model import LogisticRegression

# -------- CLIENT 1 DATA --------
X1 = [[1], [2], [3]]
y1 = [0, 0, 1]

# -------- CLIENT 2 DATA --------
X2 = [[2], [3], [4]]
y2 = [0, 1, 1]

# -------- LOCAL TRAINING --------
client1 = LogisticRegression()
client1.fit(X1, y1)

client2 = LogisticRegression()
client2.fit(X2, y2)

# -------- GET MODEL WEIGHTS --------
w1 = client1.coef_[0][0]
w2 = client2.coef_[0][0]

# -------- ADD DIFFERENTIAL PRIVACY NOISE --------
noise1 = np.random.normal(0, 0.1)
noise2 = np.random.normal(0, 0.1)

private_w1 = w1 + noise1
private_w2 = w2 + noise2

print("Client 1 Private Weight:", private_w1)
print("Client 2 Private Weight:", private_w2)

# -------- SERVER AGGREGATION --------
global_weight = (private_w1 + private_w2) / 2

print("Global Weight:", global_weight)
```
