```python
# Simple Federated Learning for Diabetes Prediction

from sklearn.linear_model import LogisticRegression
import numpy as np

# -------- CLIENT 1 DATA --------
X1 = [[1, 20],
      [2, 25],
      [3, 30]]

y1 = [0, 0, 1]

# -------- CLIENT 2 DATA --------
X2 = [[2, 22],
      [3, 28],
      [4, 35]]

y2 = [0, 1, 1]

# -------- LOCAL TRAINING --------
client1 = LogisticRegression()
client1.fit(X1, y1)

client2 = LogisticRegression()
client2.fit(X2, y2)

# -------- GET CLIENT WEIGHTS --------
w1 = client1.coef_[0]
w2 = client2.coef_[0]

print("Client 1 Weights:", w1)
print("Client 2 Weights:", w2)

# -------- FEDERATED AVERAGING --------
global_weights = (w1 + w2) / 2

print("Global Weights:", global_weights)

# -------- ACCURACY --------
acc1 = client1.score(X1, y1)
acc2 = client2.score(X2, y2)

global_accuracy = (acc1 + acc2) / 2

print("Global Accuracy:", global_accuracy)
```
