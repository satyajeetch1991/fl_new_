```python
# Simple Client-Server Federated Learning

import numpy as np
from sklearn.linear_model import LogisticRegression

# -------- CLIENT 1 DATA --------
X1 = [[2, 70],
      [3, 80],
      [4, 90]]

y1 = [0, 1, 1]

# -------- CLIENT 2 DATA --------
X2 = [[1, 60],
      [2, 65],
      [5, 95]]

y2 = [0, 0, 1]

# -------- LOCAL TRAINING --------
client1 = LogisticRegression()
client1.fit(X1, y1)

client2 = LogisticRegression()
client2.fit(X2, y2)

# -------- SEND PARAMETERS TO SERVER --------
w1 = client1.coef_[0]
w2 = client2.coef_[0]

print("Client 1 Weights:", w1)
print("Client 2 Weights:", w2)

# -------- SERVER AGGREGATION --------
global_weights = (w1 + w2) / 2

print("Updated Global Weights:", global_weights)

# -------- GLOBAL ACCURACY --------
acc1 = client1.score(X1, y1)
acc2 = client2.score(X2, y2)

global_accuracy = (acc1 + acc2) / 2

print("Global Accuracy:", global_accuracy)
```
