```python
# Basic Federated Reinforcement Learning Example

import numpy as np
import random

# -------- SIMPLE GRID --------
actions = ["UP", "DOWN", "LEFT", "RIGHT"]

# -------- AGENT 1 Q-TABLE --------
agent1_q = np.zeros(4)

# -------- AGENT 2 Q-TABLE --------
agent2_q = np.zeros(4)

# -------- LOCAL TRAINING --------
for i in range(5):

    # Agent 1 learns
    reward1 = random.randint(1, 10)
    action1 = random.randint(0, 3)
    agent1_q[action1] += reward1

    # Agent 2 learns
    reward2 = random.randint(1, 10)
    action2 = random.randint(0, 3)
    agent2_q[action2] += reward2

print("Agent 1 Q-values:", agent1_q)
print("Agent 2 Q-values:", agent2_q)

# -------- SERVER AGGREGATION --------
global_q = (agent1_q + agent2_q) / 2

print("Global Q-values:", global_q)

# -------- BEST ACTION --------
best_action = np.argmax(global_q)

print("Best Action:", actions[best_action])
```
