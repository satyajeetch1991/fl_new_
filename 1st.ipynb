```python
# Simple Federated Learning Example
import numpy as np

# -------- CLIENT --------
class Client:
    def __init__(self, data, label):
        self.data = data
        self.label = label
        self.weight = np.random.rand()

    def train(self):
        # Simple local training
        prediction = self.data * self.weight
        error = prediction - self.label

        # Update weight
        self.weight = self.weight - 0.1 * error.mean()

        return self.weight


# -------- SERVER --------
class Server:
    def aggregate(self, weights):
        # Average all client weights
        return np.mean(weights)


# -------- MAIN --------
# Client data
client1 = Client(np.array([1, 2, 3]), np.array([2, 4, 6]))
client2 = Client(np.array([2, 3, 4]), np.array([4, 6, 8]))

# Local training
w1 = client1.train()
w2 = client2.train()

print("Client 1 Weight:", w1)
print("Client 2 Weight:", w2)

# Server aggregation
server = Server()
global_weight = server.aggregate([w1, w2])

print("Global Model Weight:", global_weight)
```
