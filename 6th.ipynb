```python
# Simple Vertical Federated Learning Example

import numpy as np
from sklearn.linear_model import LogisticRegression

# -------- CLIENT A FEATURES --------
XA = np.array([[1, 2],
               [2, 3],
               [3, 4],
               [4, 5]])

# -------- CLIENT B FEATURES --------
XB = np.array([[5, 6],
               [6, 7],
               [7, 8],
               [8, 9]])

# Labels
y = np.array([0, 0, 1, 1])

# -------- COMBINE FEATURES --------
X = np.concatenate((XA, XB), axis=1)

# -------- TRAIN MODEL --------
model = LogisticRegression()
model.fit(X, y)

# -------- PREDICTION --------
pred = model.predict(X)

print("Predictions:", pred)

# -------- ACCURACY --------
accuracy = model.score(X, y)

print("Accuracy:", accuracy)
```
